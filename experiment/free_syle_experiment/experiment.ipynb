## Extract audio từ video singing

Extract file `.wav` (16 kHz, mono, PCM 16-bit) từ `data/diarization_test_set/video/singing.mp4`  
→ lưu vào `data/diarization_test_set/audio/singing.wav`

```python
import os
import subprocess

# Đường dẫn gốc của project (2 cấp trên folder này)
PROJECT_ROOT = os.path.abspath(os.path.join(os.path.dirname(os.path.abspath("__file__")), "..", ".."))

INPUT_VIDEO  = os.path.join(PROJECT_ROOT, "data", "diarization_test_set", "video",  "singing.mp4")
OUTPUT_AUDIO = os.path.join(PROJECT_ROOT, "data", "diarization_test_set", "audio", "singing.wav")

os.makedirs(os.path.dirname(OUTPUT_AUDIO), exist_ok=True)

print(f"Input  : {INPUT_VIDEO}")
print(f"Output : {OUTPUT_AUDIO}")
print(f"Input exists: {os.path.exists(INPUT_VIDEO)}")
```

```python
cmd = [
    "ffmpeg",
    "-y",                    # overwrite nếu đã tồn tại
    "-i", INPUT_VIDEO,
    "-vn",                   # bỏ video stream
    "-acodec", "pcm_s16le",  # PCM 16-bit little-endian
    "-ar", "16000",          # sample rate 16 kHz
    "-ac", "1",              # mono
    OUTPUT_AUDIO,
]

print("Chạy lệnh:")
print(" ".join(cmd))
print()

result = subprocess.run(cmd, capture_output=True, text=True)

if result.returncode == 0:
    size_mb = os.path.getsize(OUTPUT_AUDIO) / 1024 / 1024
    print(f"✅ Đã lưu: {OUTPUT_AUDIO}  ({size_mb:.2f} MB)")
else:
    print("❌ ffmpeg thất bại:")
    print(result.stderr)
```
